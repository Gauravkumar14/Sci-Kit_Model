```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

# 1️⃣ Create a fresh dataset
data = {
    "Review": [
        "I loved this movie, it was fantastic!",
        "The plot was dull and boring.",
        "Amazing cinematography and great acting.",
        "This was the worst film I've ever seen.",
        "Absolutely brilliant! A must-watch.",
        "Terrible script and horrible acting.",
        "An enjoyable movie with a great story.",
        "Not worth the time. Completely disappointed.",
        "A masterpiece! Great direction and screenplay.",
        "I regret watching this, so bad!"
    ],
    "Sentiment": ["Positive", "Negative", "Positive", "Negative", "Positive",
                  "Negative", "Positive", "Negative", "Positive", "Negative"]
}

df = pd.DataFrame(data)

# 2️⃣ Convert Sentiment to Numeric (0 = Negative, 1 = Positive)
df["Sentiment"] = df["Sentiment"].map({"Negative": 0, "Positive": 1})

# 3️⃣ Split into train and test sets
X_train, X_test, y_train, y_test = train_test_split(df["Review"], df["Sentiment"], test_size=0.2, random_state=42)

# 4️⃣ Convert text to numerical representation (TF-IDF)
vectorizer = TfidfVectorizer(stop_words="english", ngram_range=(1, 2))
X_train_vectorized = vectorizer.fit_transform(X_train)
X_test_vectorized = vectorizer.transform(X_test)

# 5️⃣ Train Naive Bayes Model
model = MultinomialNB()
model.fit(X_train_vectorized, y_train)

# 6️⃣ Predict on test data
y_pred = model.predict(X_test_vectorized)

# 7️⃣ Evaluate Model Performance
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy:.4f}\n")
print("Classification Report:\n", classification_report(y_test, y_pred))
```
